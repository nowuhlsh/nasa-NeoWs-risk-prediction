## Notebook 1: Data collection and exploration
I began by producing functions capable of extracting API data for chosen time periods, and packaging the raw JSON data into pandas dataframes. These functions sit in *api.py*. This meant that I was able to manipulate, clean, and validate the data before inserting it into an SQLite database. By doing so, I had a permanent historical record of close approach events.

```python
import pandas as pd
from src.api import asteroid_data_extract, rec_to_df, fetch_dates_asteroids, fetch_daterange_asteroids
from src.processing import standardise_asteroid_data, asteroid_data_report
```

```python
#checking functions so far
asteroids = fetch_dates_asteroids("2026-09-12")
rec = asteroid_data_extract(asteroids)
df = rec_to_df(rec)
print(df.dtypes)
standardised_df = standardise_asteroid_data(df)
print(standardised_df.dtypes)
```

Then, I tested code that I wanted to put in my validating function. I wanted to be able to check for missing data, duplicated data, and then standardise the format of the data.

```python
#Checking for missing data
missing = standardised_df.isna().sum()
print(missing)
print(standardised_df)
```

Also, my original function could not handle dates that produced no close approach events, so I introduced code that could handles this into the *fetch_dates_asteroids* function.
